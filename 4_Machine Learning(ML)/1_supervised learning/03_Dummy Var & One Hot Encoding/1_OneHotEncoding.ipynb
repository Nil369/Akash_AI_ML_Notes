### ***2) Label Encoder & One Hot Encoder***

- ***Now use one hot encoder to create dummy variables for each of the town***

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
ct = ColumnTransformer([('town', OneHotEncoder(), [0])], remainder = 'passthrough')
```

```python
X = ct.fit_transform(X)
X
```

```python
X = X[:,1:]
print(X)
```

```python
from sklearn import linear_model
model = linear_model.LinearRegression()
```

```python
model.fit(X,y)
```

```python
model.predict([[0,1,3400]]) # 3400 sqr ft home in west windsor
```

```python
model.predict([[1,0,2800]]) # 2800 sqr ft home in robbinsville
```

```python
model.score(X,y)*100
```

### <font style="color:darkorange;font-weight:700;">Diff. Types of Encoding:</font>

1. ***Label Encoding*** is used when **`categories are nominal (no order) and converts them into integers.`**
2. ***One-Hot Encoding*** is used to **`create binary columns for each category`, useful when there is `no ordinal relationship`**.
3. ***Ordinal Encoding*** is used for **`categorical features with a meaningful order`**.

`Example Scenario:`
Imagine we have a dataset of fruits with the following information:

- **Fruit**: Names of the fruits (categorical feature)
- **Size**: The size of the fruit (ordinal feature — small, medium, large)
- **Color**: The color of the fruit (categorical feature)


```python
# Sample data
data = {
    'Fruit': ['Apple', 'Banana', 'Cherry', 'Apple', 'Cherry'],
    'Size': ['Small', 'Medium', 'Large', 'Small', 'Large'],
    'Color': ['Red', 'Yellow', 'Red', 'Green', 'Red']
}

df = pd.DataFrame(data)
df
```

**1. Label Encoding (for the Fruit column)**\
Label Encoding assigns a unique integer to each category. Let’s apply Label Encoding to the Fruit column:

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['Fruit_LabelEncoded'] = le.fit_transform(df['Fruit'])
df
```

**2. One-Hot Encoding (for the Color column)**\
One-Hot Encoding creates binary columns for each category. Let’s apply One-Hot Encoding to the Color column:

```python
# One-Hot Encoding for 'Color'
ohe = OneHotEncoder(sparse_output=False)  # Ensure it returns a dense array
color_encoded = ohe.fit_transform(df[['Color']])

# Create new columns for one-hot encoded 'Color'
color_columns = ohe.get_feature_names_out(['Color'])
df_color_encoded = pd.DataFrame(color_encoded, columns=color_columns)

# Concatenate original DataFrame with one-hot encoded columns
df_final = pd.concat([df, df_color_encoded], axis=1)

df_final
```

**3. Ordinal Encoding (for the Size column)**\
- Ordinal Encoding assigns integers based on the order of categories. 
- Since Size has a natural order (Small < Medium < Large), we can encode it as:

```python
from sklearn.preprocessing import OrdinalEncoder

size_order = ['Small', 'Medium', 'Large']
ordinal_encoder = OrdinalEncoder(categories=[size_order])
df['Size_OrdinalEncoded'] = ordinal_encoder.fit_transform(df[['Size']]).astype(int)
df
```

***similary it can be done using map function:***

```python
size_mapping = {'Small': 0, 'Medium': 1, 'Large': 2}
df['Size_OE_en_map'] = df['Size'].map(size_mapping)
df
```
