<h1 style="color:tomato; text-align:center; font-weight:700">Group by in Pandas</h1>

```python
import pandas as pd
df = pd.read_csv("weather_by_cities.csv")
df
```

### For this dataset, we are going to get the following answers,
***1. What was the maximum temperature in each of these 3 cities?***<br>
***2. What was the average windspeed in each of these 3 cities?***

```python
grouped_city = df.groupby('city')
grouped_city
```

```python
for city, data in grouped_city:
    print("city:",city)
    print("\n")
    print("data:",data) 
```

<h3>
Group data using custom function: Let's say you want to group your data using custom function. Here the requirement is to create three groups<br><br>
1. Days when temperature was between 80 and 90<br>
2. Days when it was between 50 and 60<br>
3. Days when it was anything else
</h3>

### For this you need to write custom grouping function and pass that to groupby

```python
def grouper(df, idx, col):
    if 80 <= df[col].loc[idx] <= 90:
        return '80-90'
    elif 50 <= df[col].loc[idx] <= 60:
        return '50-60'
    else:
        return 'others'
```

```python
g = df.groupby(lambda x: grouper(df, x, 'temperature'))
g
```

```python
for key, d in g:
    print("Group by Key: {}\n".format(key))
    print(d)
```
