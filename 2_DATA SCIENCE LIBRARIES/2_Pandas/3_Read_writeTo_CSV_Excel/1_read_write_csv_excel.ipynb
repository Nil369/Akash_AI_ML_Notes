<h3 style="color:blue; font-weight:700">Read Excel</h3>

```python
data = pd.read_excel("stock_data.xlsx")
data
```

```python
def convert_people_cell(cell):
    if cell=="n.a.":
        return 'Sam Walton'
    return cell

def convert_price_cell(cell):
    if cell=="n.a.":
        return 50
    return cell
    
df = pd.read_excel("stock_data.xlsx","Sheet1", converters= {
        'people': convert_people_cell, # Remeber don't execute the function just give the reference of it
        'price': convert_price_cell
    })
df
```

<h3 style="color:blue; font-weight:700">Write Excel</h3>

```python
df.to_excel("new_stock.xlsx", sheet_name="stocks", index=False, startrow=2, startcol=1)
```
