<h1 style="color:orange;">DATA FRAMES</h1>

- Dataframes is a ***main object*** in **Pandas**.
- It is used to `represent data with rows & columns (tabular/excel spreadsheet)`

```python
import pandas as pd
weather_data = {
    'day': ['1/1/2017','1/2/2017','1/3/2017','1/4/2017','1/5/2017','1/6/2017'],
    'temperature': [32,35,28,24,32,31],
    'windspeed': [6,7,2,7,4,2],
    'event': ['Rain', 'Sunny', 'Snow','Snow','Rain', 'Sunny']
}
df = pd.DataFrame(weather_data)
df = pd.read_csv("weather_data.csv")
df
```

<h2 style="color:blue;">Operations on Data frames</h2>

```python
df['temperature'].max() #df.temperature.max()
```

```python
df[df['temperature']>=32]
```

```python
df['day'][df['temperature'] == df['temperature'].max()] # Kinda doing SQL in pandas
```

```python
df[df['temperature'] == df['temperature'].max()] #Extracting the data frame by quering like SQL in pandas
```

```python
df['temperature'].std()
```

```python
df['event'].max() # But mean() won't work since data type is string
```

```python
df.describe() #For printing all the statistics
```
