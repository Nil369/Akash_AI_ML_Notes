<h1 style="color:blueviolet; font-weight:700; text-align:center">Mean, Median, Mode, Percentile</h1>

### 1. `Mean`:The mean is the ***average of a dataset***.

Example\
**Given data**: `[10, 20, 30, 40, 50]`\
**Mean** = (10 + 20 + 30 + 40 + 50) / 5 = 30

```python
import numpy as np

data = [10,20,30,40,50]
mean = np.mean(data)
print(f"Mean: {int(mean)}")
```

### 2. `Median`: The median is the ***middle value*** when the data is sorted.

```python
import pandas as pd

df = pd.read_csv("income.csv", names=["name","income"], skiprows=[0])
df
```

- In the dataset we can clearly see that taking the `mean for the data set will give an terrible average!`
- As Mr. Elon Musk is an outlier here...Because of which the monthly average would be around **1.43 million**
- `Outlier` is a dataset that is ***very different from the rest of the data points***
- In this situation taking a **`median would be a much better option to remove the outlier`**

```python
# print the stats for the whole income column like this:
df.income.describe()
```

- ***The percentile tells you what percentage of data falls below a certain value.***
- In pandas perecntile == quantile. It tells `what % of data falls below a certain value`

```python
print(df["income"].quantile(0))
```

```python
print(df.income.quantile(0.25)) # by default it will consider the average btw lower & higher bound in a median
```

```python
print(df.income.quantile(0.25,interpolation="higher")) # it will consider the upper/higher bound in a median
```

```python
print(df.income.quantile(0.25,interpolation="lower")) # it will consider the lower bound in a median
```

```python
print(df.income.quantile(0.75))
```

```python
print(df.income.quantile(1))
```

```python
percentile_99 = df.income.quantile(0.99)
print(percentile_99)
```

```python
df[df.income>percentile_99]
```

```python
df.loc[3, 'income'] = np.nan
df
```

<h4>• Now see using mean gives an horrible average</h4>

```python
print(df.income.mean())
```

```python
df_new = df.fillna(df.income.mean())
df_new
```

<h4>• Now using median gives better result</h4>

```python
df_new = df.fillna(df.income.median())
df_new
```

### 3. `Mode`: value that appears most frequently in a dataset

Example:\
Given data: `[10, 20, 20, 30, 40, 50]`\
Mode = 20 (since it appears twice)

```python
data_with_mode = [10, 20, 20, 30, 40, 50]

values, counts = np.unique(data_with_mode, return_counts=True)
mode = values[np.argmax(counts)]

print("Mode:", mode)
```
