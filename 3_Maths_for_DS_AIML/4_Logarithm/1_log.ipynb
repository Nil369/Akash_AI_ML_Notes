## What is a Logarithm?
A logarithm answers the question: ***`"To what exponent must a base be raised to produce a given number?"`***

```python
import pandas as pd

df = pd.read_csv("revenue.csv")
df
```

```python
df.plot(x='company', y='revenue', kind='bar',rot=45)
```

```python
df.plot(x='company', y='revenue',kind='bar', logy=True,rot=40)
```
