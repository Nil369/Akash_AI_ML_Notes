```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
import numpy as np
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt
digits = load_digits()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(digits.data,digits.target,test_size=0.3)
```

**Logistic Regression**

```python
lr = LogisticRegression(solver='liblinear')
lr.fit(X_train, y_train)
lr.score(X_test, y_test)*100
```

**SVM**

```python
svm = SVC(gamma='auto')
svm.fit(X_train, y_train)
svm.score(X_test, y_test)*100
```

**Random Forest**

```python
rf = RandomForestClassifier(n_estimators=40)
rf.fit(X_train, y_train)
rf.score(X_test, y_test)*100
```

<h2 style='color:tomato'>KFold cross validation</h2>

**Basic example**

```python
from sklearn.model_selection import KFold
kf = KFold(n_splits=3)
kf
```

```python
for train_index, test_index in kf.split([1,2,3,4,5,6,7,8,9]):
    print(train_index, test_index)
```

**Use KFold for our digits example**

```python
def get_score(model, X_train, X_test, y_train, y_test):
    model.fit(X_train, y_train)
    return model.score(X_test, y_test)
```

```python
from sklearn.model_selection import StratifiedKFold
folds = StratifiedKFold(n_splits=3)

scores_logistic = []
scores_svm = []
scores_rf = []

for train_index, test_index in folds.split(digits.data,digits.target):
    X_train, X_test, y_train, y_test = digits.data[train_index], digits.data[test_index], \
                                       digits.target[train_index], digits.target[test_index]
    scores_logistic.append(get_score(LogisticRegression(solver='liblinear'), X_train, X_test, y_train, y_test))  
    scores_svm.append(get_score(SVC(gamma='auto'), X_train, X_test, y_train, y_test))
    scores_rf.append(get_score(RandomForestClassifier(n_estimators=40), X_train, X_test, y_train, y_test))
```

```python
[x * 100 for x in scores_logistic]
```

```python
[x * 100 for x in scores_svm]
```

```python
[x*100 for x in scores_rf]
```

<h2 style='color:tomato'>cross_val_score function</h2>

```python
from sklearn.model_selection import cross_val_score
```

**Logistic regression model performance using cross_val_score**

```python
cross_val_score(LogisticRegression(solver='liblinear'), digits.data, digits.target,cv=3)
```

**svm model performance using cross_val_score**

```python
cross_val_score(SVC(gamma='auto'), digits.data, digits.target,cv=3)*100
```

**random forest performance using cross_val_score**

```python
cross_val_score(RandomForestClassifier(n_estimators=40),digits.data, digits.target,cv=3)*100
```

***cross_val_score uses stratifield kfold by default***

<h2 style='color:tomato'>Parameter tunning using k fold cross validation</h2>

```python
scores1 = cross_val_score(RandomForestClassifier(n_estimators=5),digits.data, digits.target, cv=10)
print(np.average(scores1))
```

```python
scores2 = cross_val_score(RandomForestClassifier(n_estimators=20),digits.data, digits.target, cv=10)
print(np.average(scores2)*100)
```

```python
scores3 = cross_val_score(RandomForestClassifier(n_estimators=30),digits.data, digits.target, cv=10)
print(np.average(scores3)*100)
```

```python
scores4 = cross_val_score(RandomForestClassifier(n_estimators=40),digits.data, digits.target, cv=10)
print(np.average(scores4)*100)
```

Here we used cross_val_score to
fine tune our random forest classifier and figured that having around 40 trees in random forest gives best result.
