<h1 style="color:green;font-weight:700;text-align:center;">DBSCAN</h1>

---

**DBSCAN (Density-Based Spatial Clustering of Applications with Noise)** is an unsupervised learning algorithm that **groups points based on their density**:

- ***`Core Points`***: **Points with at least min_samples neighbors within a radius eps.**
- ***`Border Points`***: **Points that are reachable from core points but have fewer than min_samples neighbors.**
- ***`Outliers`***: **Points that are neither core nor border points.**

***DBSCAN does not require specifying the number of clusters beforehand and can handle noise well.***


```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
```

```python
x,y = make_moons(n_samples=250,noise=0.05,random_state=42)
```

```python
dataset = {"data1":x[:,0],"data2":x[:,1],"output":y}

df = pd.DataFrame(dataset)
df.head(3)
```

```python
import seaborn as sns

sns.scatterplot(x="data1",y="data2",data=df)
plt.show()
```

## DBSCAN Clusterring

```python
from sklearn.cluster import DBSCAN

db = DBSCAN(eps=0.3, min_samples=5)
df["Predict"] = db.fit_predict(df)
```

```python
df.head()
```

```python
sns.scatterplot(x="data1",y="data2",data=df,hue="Predict")
plt.show()
```
