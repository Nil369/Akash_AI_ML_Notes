<h1 style="color:blueviolet;font-weight:700;text-align:center;">Apriori Algorithm</h1>

---

The Apriori algorithm is used for mining frequent itemsets and generating **association rules**. It uses a **bottom-up approach** where:

1. Frequent individual items are identified.
2. Larger itemsets are formed by combining smaller itemsets.
3. The process stops when no more frequent itemsets can be formed.

#### **`Steps`**:

- ***Set a minimum support threshold.***
- ***Generate candidate itemsets and filter based on support.***
- ***Continue until no candidate meets the threshold.***

```python
# !pip install mlxtend
```

```python
from mlxtend.frequent_patterns import apriori, association_rules
import pandas as pd
```

```python
data = {
    'Transaction': [
        ['Milk', 'Bread', 'Eggs', 'Butter'],
        ['Milk', 'Diapers', 'Beer'],
        ['Milk', 'Bread', 'Butter'],
        ['Bread', 'Eggs'],
        ['Diapers', 'Beer', 'Bread'],
        ['Bread', 'Milk', 'Butter'],
        ['Milk', 'Eggs'],
        ['Diapers', 'Eggs', 'Beer', 'Bread'],
        ['Milk', 'Bread', 'Butter'],
        ['Diapers', 'Beer'],
        ['Eggs', 'Bread'],
        ['Milk', 'Bread', 'Butter', 'Eggs'],
        ['Diapers', 'Milk', 'Beer'],
        ['Milk', 'Eggs', 'Bread'],
        ['Milk', 'Butter', 'Bread', 'Diapers']
    ]
}
```

```python
# Extract unique items
all_items = sorted(set(item for transaction in data['Transaction'] for item in transaction))

# Convert dataset into a one-hot encoded DataFrame
transactions = pd.DataFrame([{item: (item in transaction) for item in all_items} for transaction in data['Transaction']])
```

```python
# Apply Apriori algorithm with a minimum support of 0.5
frequent_itemsets = apriori(transactions, min_support=0.5, use_colnames=True)
```

```python
print("Frequent Itemsets:\n", frequent_itemsets)
```

---
### ***Frequent Pattern Growth (FP-Growth) Algorithm***

FP-Growth is an improvement over Apriori. Instead of generating candidate itemsets explicitly, it constructs a FP-tree (Frequent Pattern Tree) to encode transactions and extract frequent itemsets efficiently.

```python
from mlxtend.frequent_patterns import fpgrowth

# Apply FP-Growth algorithm
frequent_itemsets_fp = fpgrowth(transactions, min_support=0.6, use_colnames=True)

print("Frequent Itemsets using FP-Growth:\n", frequent_itemsets_fp)
```
