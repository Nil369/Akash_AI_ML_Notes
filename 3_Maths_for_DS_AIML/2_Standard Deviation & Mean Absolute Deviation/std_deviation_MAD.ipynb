## <font style="color:#FF362E;">Standard Deviation<font>
Standard deviation **`measures how spread out the values are from the mean`**.

Example:
- Given data: [10, 20, 30, 40, 50]
- The standard deviation is calculated using:

<img src="img/std_formula.png" alt="std::formula">

```python
import numpy as np
```

```python
data = [10, 20, 30, 40, 50]
std_dev = np.std(data)
print("Standard Deviation:", std_dev)
```

## <font style="color:#45CE30;">Mean Absolute Deviation (MAD)</font>
MAD measures the **`average distance between each data point`** and the `mean`.

```python
mean = np.mean(data)
mad = np.mean(np.abs(data - mean))
print("Mean Absolute Deviation:", mad)
```
