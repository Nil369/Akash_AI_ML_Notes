#### **Probelem Statement**:
***Predict canada's per capita income in year 2020. Using the provided csv build a regression model and predict the per capita income fo canadian citizens in year 2020***

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df = pd.read_csv("canada_per_capita_income.csv")
df.head()
```

```python
plt.xlabel("Year")
plt.xlabel("Capital Income")
plt.title("Canada Capital Income")
plt.scatter(df.year,df["per capita income (US$)"], marker="+", color="red")
plt.show()
```

### Training the model

```python
X = df[["year"]]
y = df[["per capita income (US$)"]]
print(X.head(),y.head())
```

```python
model = linear_model.LinearRegression()
model.fit(X,y)
```

```python
import warnings
warnings.filterwarnings("ignore", category=UserWarning)
```

#### Predicting canada's per capita income in year 2020.

```python
model.predict([[2020]])
```

```python
plt.xlabel("Year")
plt.ylabel("Capital Income")
plt.title("Canada Capital Income")

# Scatter plot of actual data
plt.scatter(df.year, df["per capita income (US$)"], marker="+", color="red")

# Linear regression line
plt.plot(df.year, model.predict(df.year.values.reshape(-1,1)), color="blue")

plt.show()
```

```python
import numpy as np

year_df = np.array([2011,2012,2013,2014,2015,2016,2017,2018,2019,2020,2021,2022,2023,2024,2025])
year_df
```

```python
# Convert it to a pandas DataFrame
year_df = pd.DataFrame(year_df, columns=['year'])  # Pass the column name as a list
year_df.head()
```

```python
predictions = model.predict(year_df)
predictions
```

```python
# Assign predictions to the column
year_df["predicted capital income"] = predictions

# Drop the column with the correct name (no colon)
year_df.drop('predicted capital income', axis="columns", inplace=True)

# Print the last 6 rows of the DataFrame
print(year_df.tail(6))
```

```python
year_df.to_csv("predicted_canadian_capital_income.csv",index=False)
```
