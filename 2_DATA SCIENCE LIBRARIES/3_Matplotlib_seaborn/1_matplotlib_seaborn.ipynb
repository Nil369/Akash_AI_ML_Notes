#### pyplot.plot format strings and other options

```python
import matplotlib.pyplot as plt
%matplotlib inline

x=[1,2,3,4,5,6,7]
y=[50,51,52,48,47,49,46]
```

```python
plt.xlabel('Day')
plt.ylabel('Temperature')
plt.title('Weather')

plt.plot(x,y,'b+--')
```

```python
plt.plot(x,y,'g<',alpha=0.5) # alpha can be specified on a scale 0 to 1
```

<h2 style="color:tomato; font-weight:600;"> Reading from excel files to plot graphs/charts</h2>

```python
import pandas as pd

df_sales = pd.read_excel("linechart.xlsx")
df_sales
```

## Line Chart

```python
from matplotlib import pyplot as plt

plt.figure(figsize=(12,4)) # Size of the figure
plt.plot(df_sales['Quarter'], df_sales['Fridge'],color="blue",label="Fridge")
plt.plot(df_sales['Quarter'], df_sales['Dishwasher'],color="green",label="Dishwasher")
plt.plot(df_sales['Quarter'], df_sales['Washing Machine'],color="orange",label="Washing Machine")

plt.title("Product Sales")
plt.ylabel("Revenue in Million $")
plt.xlabel("Financial Quarter")

plt.legend(loc='upper left',shadow=True)
# plt.grid(axis="x")

plt.show()
```

## Pie Chart

```python
total_sales = df_sales[["Fridge","Dishwasher","Washing Machine"]].sum()
total_sales
```

```python
plt.pie(total_sales, labels=total_sales.index, autopct='%1.2f%%', startangle=140, explode=(0.1,0,0), shadow=True)
plt.show()
```

### Saving Pie Charts/ any chart/graph using savefig

```python
exp_vals = [1400,600,300,410,250]
exp_labels = ["Home Rent","Food","Phone/Internet Bill","Car ","Other Utilities"]

plt.pie(exp_vals,labels=exp_labels,radius=2, autopct='%0.1f%%', explode=[0,0.1,0.1,0,0])
plt.savefig("pie_chart.jpg", bbox_inches="tight", pad_inches=2, transparent=True)
```

## Bar Chart

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
```

### Simple bar chart showing revenues of major US tech companies

```python
company=['GOOGL','AMZN','MSFT','FB']
revenue=[90,136,89,27]
xpos = np.arange(len(company))
xpos
```

```python
plt.bar(xpos,revenue, label="Revenue")

plt.xticks(xpos,company)
plt.ylabel("Revenue(Bln)")
plt.title('US Technology Stocks')
plt.legend()
```

### Multiple Bars showing revenue and profit of major US tech companies

```python
profit=[40,2,34,12]

plt.bar(xpos-0.2,revenue, width=0.4, label="Revenue")
plt.bar(xpos+0.2,profit, width=0.4,label="Profit")

plt.xticks(xpos,company)
plt.ylabel("Revenue(Bln)")
plt.title('US Technology Stocks')
plt.legend()
```

### Horizontal bar chart using barh function

```python
plt.barh(xpos,revenue, label="Revenue")

plt.yticks(xpos,company)
plt.ylabel("Revenue(Bln)")
plt.title('US Technolog Stocks')
plt.legend()
```

## Histogram

**In histograms, x axis contains a variable and y axis will be a frequency of that variable**

**We have a sample data of blood sugar level of different patients, we will try to plot number of patients by blood range and try to figure out how many patients are normal, pre-diabetic and diabetic**

```python
blood_sugar = [113, 85, 90, 150, 149, 88, 93, 115, 135, 80, 77, 82, 129]
plt.hist(blood_sugar, rwidth=0.8) # by default number of bins is set to 10
plt.show()
```

```python
plt.hist(blood_sugar,rwidth=0.5,bins=4)
plt.show()
```

<h3 style='color:purple'>Histogram showing normal, pre-diabetic and diabetic patients distribution</h3>
<ul style="color:brown;font-weight: bold;">
    <li>80-100: Normal</li>
    <li>100-125: Pre-diabetic</li>
    <li>80-100: Diabetic</li>
</ul>

```python
plt.xlabel("Sugar Level")
plt.ylabel("Number Of Patients")
plt.title("Blood Sugar Chart")

plt.hist(blood_sugar, bins=[80,100,125,150], rwidth=0.95, color='g')
plt.show()
```

<h3 style='color:purple'>Mutiple data samples in a histogram</h3>

```python
plt.xlabel("Sugar Level")
plt.ylabel("Number Of Patients")
plt.title("Blood Sugar Chart")

blood_sugar_men = [113, 85, 90, 150, 149, 88, 93, 115, 135, 80, 77, 82, 129]
blood_sugar_women = [67, 98, 89, 120, 133, 150, 84, 69, 89, 79, 120, 112, 100]

plt.hist([blood_sugar_men,blood_sugar_women], bins=[80,100,125,150], rwidth=0.95, color=['green','orange'],label=['men','women'])
plt.legend()
plt.show()
```

```python
plt.xlabel("Sugar Level")
plt.ylabel("Number Of Patients")
plt.title("Blood Sugar Chart")

plt.hist(blood_sugar,bins=[80,100,125,150],rwidth=0.95,histtype='step')
```

```python
plt.xlabel("Number Of Patients")
plt.ylabel("Sugar Level")
plt.title("Blood Sugar Chart")

plt.hist(blood_sugar, bins=[80,100,125,150], rwidth=0.95, orientation='horizontal')
plt.show()
```

## Bubble Chart

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

data = pd.read_csv('https://raw.githubusercontent.com/holtzy/The-Python-Graph-Gallery/master/static/data/gapminderData.csv')
data['continent'] = pd.Categorical(data['continent'])
data.head()
```

```python
fig, ax = plt.subplots(figsize=(10, 10))

# Subset of the data for year 1952
data1952 = data[data.year == 1952]

# Ensure `continent` column is categorical if it isn't already
if not isinstance(data1952['continent'].dtype, pd.CategoricalDtype):
    data1952['continent'] = data1952['continent'].astype('category')

# Scatter plot for each continent with labels
continents = data1952['continent'].cat.categories  # Extract unique categories
for continent in continents:
    subset = data1952[data1952['continent'] == continent]
    ax.scatter(
        x=subset['lifeExp'],
        y=subset['gdpPercap'],
        s=subset['pop'] / 50000,
        label=continent,  # Add continent as the label
        alpha=0.6,
        edgecolors="white",
        linewidth=2,
    )

# Add titles (main and on axis)
plt.yscale('log')
ax.set_xlabel("Life Expectancy")
ax.set_ylabel("GDP per Capita")
ax.set_title("Year 1952")
ax.set_ylim(100, 50000)
ax.set_xlim(30, 75)

ax.legend(title="Continents", loc="upper left")

plt.show()
```
