<h1 style="color:darkorange;font-weight:700;text-align:center">Naive Bayes</h1>

---
- We know Bayes Theorem:
  ```P(A|B) = P(B|A).P(A)/P(B)```
  
- We call this **`Naive Bayes`** because we ***make a naive assumption on the features that they will be independent variables. Like if we drop them also it will not effect the main goal => to make accurate prediction for the problem satatement.***

````python
import pandas as pd

df = pd.read_csv("titanic.csv")
df[:5]
````

***Here we are making naive assumptions & dropping some columns assuming it will not affect the survival on these columns***

````python
df.drop(['PassengerId','Name','SibSp','Parch','Ticket','Cabin','Embarked'],axis='columns',inplace=True)
df.head()
````

````python
inputs = df.drop('Survived',axis="columns")
target = df.Survived
````

````python
inputs.Sex
````

````python
dummies = pd.get_dummies(inputs.Sex) 

# Convert the boolean values in the dummy variables to 1 and 0 explicitly
dummies = dummies.astype(int)
dummies.head()
````

````python
inputs = pd.concat([inputs,dummies],axis="columns")
inputs.head()
````

````python
inputs.drop(['Sex','male'],axis='columns',inplace=True)
inputs.head(3)
````

````python
inputs.columns[inputs.isna().any()]
````

````python
inputs.Age[:10]
````

````python
inputs.Age = inputs.Age.fillna(inputs.Age.mean())
inputs.head()
````

````python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(inputs,target,test_size=0.3)
````

````python
len(X_train)
````

````python
len(X_test)
````

````python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
````

````python
model.fit(X_train,y_train)
````

````python
model.score(X_test,y_test)
````
