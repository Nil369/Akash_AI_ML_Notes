# Pivot Table

A pivot table is a **table of values** which are `aggregations of groups of individual values` from a more extensive table (such as from a database, spreadsheet, or business intelligence program) within one or more discrete categories.

```python
df = pd.read_csv("weather2.csv")
df
```

```python
df.pivot_table(index="city",columns="date")
```

### <code style="color:blue; font-weight:800;">Margins</code>

```python
df.pivot_table(index="city",columns="date", margins=True, aggfunc="sum")
```

### <code style="color:blue; font-weight:800;">Grouper</code>

```python
df = pd.read_csv("weather3.csv")
df
```

```python
df['date'] = pd.to_datetime(df['date'])
df.pivot_table(index=pd.Grouper(freq='ME',key='date'),columns='city')
```

### <font style="color:salmon; font-weight:700;">Reshape pandas dataframe using melt</font>

```python
df = pd.read_csv("weather_4.csv")
df
```

```python
melted = pd.melt(df, id_vars=["day"], var_name='city', value_name='temperature')
melted
```
